```python
# Our cleaning focus is now streamlined to Education_Level, Marital_Status, Income_Category
# We keep Unknown values as a category on its own in each column but will map Income_Category column as an ordinal column
# That is, keeping the column in that order helps the model learn the pattern that category 5 is more than category 3 or 
# Category 2 is less than Category 4
```

```python
income_map = {
    "Less than $40K": 1,
    "$40K - $60K": 2,
    "$60K - $80K": 3,
    "$80K - $120K": 4,
    "$120K +": 5,
    "Unknown": 0
}

bank_churn_copy['Income_Category_New'] = bank_churn_copy['Income_Category'].map(income_map)
```

```python
# Next, split the data before encoding.
# First, separate the features (X) and target (y).
# Then split the data into training and test sets using train_test_split().
# After splitting, fit and transform the encoders on the training data only and use the
# LabelEncoder for the target variables
# Use the encoded data to fit your model (X_train for fitting and X_test for predicting)
# Do not OneHotEncode the target variable. Only use LabelEncoder for y_train and y_test
# This helps prevent data leakage from the test set into the model.
```

```python
X = bank_churn_copy.drop(columns=['Attrition_Flag'])
y = bank_churn_copy['Attrition_Flag']
```

```python
X.head(3)
```

```python
X = X.drop(columns=['Income_Category'])
```

```python
y.head(3)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
```

```python
# Now, we encode our X_train data and y_train data using OneHotEncoder and LabelEncoder respectively
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder = OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore')
```

```python
X_train
```

```python
# This would encode the train and test data but make sure never to fit the enconder with the test data
X_train_enc = encoder.fit_transform(X_train[['Education_Level', 'Marital_Status']])
X_test_enc = encoder.transform(X_test[['Education_Level', 'Marital_Status']])
```

```python
X_train_enc
```

```python
# Turn the encoded columns into a dataframe
X_train_enc_df = pd.DataFrame(X_train_enc, columns=encoder.get_feature_names_out(['Education_Level', 'Marital_Status']), 
                           index=X_train.index)
```

```python
X_train_enc_df.head(3)
```

```python
# Turn X_test to a dataframe
X_test_enc_df = pd.DataFrame(X_test_enc, columns=encoder.get_feature_names_out(['Education_Level', 'Marital_Status']),
                             index=X_test.index)
```

```python
X_test_enc_df.head(3)
```

```python
# Now, we join it back to the original dataframe
X_train_1 = pd.concat([X_train, X_train_enc_df], axis=1)
X_test_1 = pd.concat([X_test, X_test_enc_df], axis=1)
```

```python
# Now, we drop the corresponding text columns
X_train_1 = X_train_1.drop(columns=['Education_Level', 'Marital_Status'])
X_test_1 = X_test_1.drop(columns=['Education_Level', 'Marital_Status'])
```

```python
# Now, our train and test data is ready for fitting and predicting respectively
# But first, let's encode our target variable as it appears as a text column
# And scale the numerical columns with larger ranges.
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
y_train
```

```python
y_train_enc = le.fit_transform(y_train)
y_test_enc = le.transform(y_test)
```

```python
# Turn y_train_enc and y_test_enc to dataframe - not necessary but can still work fine so why not?
y_train_enc = pd.DataFrame(y_train_enc, columns=['Stay'], index=y_train.index)
y_test_enc = pd.DataFrame(y_test_enc, columns=['Stay'], index=y_test.index)
```

```python
print("X_train shape:", X_train_1.shape)
print("X_test shape: ", X_test_1.shape)
print("y_train shape:", y_train_enc.shape)
print("y_test shape: ", y_test_enc.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from xgboost import XGBClassifier

models = {
    'Logistic Regression': LogisticRegression(max_iter=5000),
    'Decision Tree': DecisionTreeClassifier(random_state=42),
    'Random Forest': RandomForestClassifier(random_state=42),
    'SVM': SVC(),
    'KNN': KNeighborsClassifier(),
    'XGB': XGBClassifier(random_state=42, eval_metric='logloss') # we use 'logloss' for binary and 'mlogloss' for multiclass
}

for name, model in models.items():
    model.fit(X_train_1, y_train_enc)
    train = model.score(X_train_1, y_train_enc)
    test = model.score(X_test_1, y_test_enc)
    gap = train - test
    print(f"{name:22s} train={train:.2f}, test={test:.2f}, gap={gap:.2f}")
```

```python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(random_state=42)
model.fit(X_train_1, y_train_enc)

y_pred = model.predict(X_test_1)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

print(confusion_matrix(y_test_enc, y_pred))
print(classification_report(y_test_enc, y_pred))
```

```python
model_balanced = RandomForestClassifier(random_state=42, class_weight='balanced')
model_balanced.fit(X_train_1, y_train_enc)
y_pred_bal = model.predict(X_test_1)

print(classification_report(y_test_enc, y_pred_bal))
```
